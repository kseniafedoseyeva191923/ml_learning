## Часть 3: Скорость — NumPy vs цикл

Посмотрим насколько NumPy быстрее обычного Python цикла.

```python
import time

# Создаём большой массив — 1 миллион точек каротажа
big_GK = np.random.uniform(20, 150, 1_000_000)

# Способ 1: Python цикл
start = time.time()
result_loop = []
for val in big_GK:
    result_loop.append(val * 2)
time_loop = time.time() - start

# Способ 2: NumPy (векторизация)
start = time.time()
result_numpy = big_GK * 2
time_numpy = time.time() - start

print(f'Цикл:  {time_loop:.4f} сек')
print(f'NumPy: {time_numpy:.4f} сек')
print(f'NumPy быстрее в {time_loop/time_numpy:.0f} раз!')
```

## Часть 4: Математика без циклов — это и есть суть NumPy

```python
GK = np.array([45, 62, 78, 55, 90, 110, 88, 42, 67, 73])

# Базовая статистика одной командой:
print('Среднее ГК:', np.mean(GK))
print('Минимум:', np.min(GK))
print('Максимум:', np.max(GK))
print('Стандартное отклонение:', np.std(GK).round(2))
```

```python
# Нормализация ГК (приведение к диапазону 0-1) — формула MinMax
# Это стандартная операция перед ML!
GK_norm = (GK - np.min(GK)) / (np.max(GK) - np.min(GK))

print('Исходный ГК:     ', GK)
print('Нормализованный: ', GK_norm.round(2))
```

## Часть 5: Глубины и срезы (slicing)

```python
# Создаём массив глубин (каждые 0.1м)
depth = np.arange(1000.0, 1001.0, 0.1)
GK    = np.array([45, 62, 78, 55, 90, 110, 88, 42, 67, 73])

print('Глубины:', depth)
print('ГК:     ', GK)

# Срез: только первые 3 точки
print('\nПервые 3 точки:')
print('  Глубина:', depth[:3])
print('  ГК:     ', GK[:3])

# Срез: последние 3 точки
print('\nПоследние 3 точки:')
print('  Глубина:', depth[-3:])
print('  ГК:     ', GK[-3:])
```

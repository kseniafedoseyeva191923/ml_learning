## Что такое Pandas?

Pandas — это библиотека для работы с **таблицами** в Python.
Думай о ней как об Excel, но:
- умеет обрабатывать **миллионы строк**
- можно автоматизировать любые операции
- идеально интегрируется с NumPy и ML-библиотеками

В геологии: твои данные каротажа — это таблица.
Строки = точки глубины, столбцы = кривые (ГК, ПС, КНД, ...).

---

## Два главных объекта Pandas:

| Объект | Что это | Аналогия |
|--------|---------|----------|
| **Series** | Один столбец (одномерный) | Одна кривая каротажа |
| **DataFrame** | Таблица (двумерная) | Весь планшет скважины |

````python
# Импортируем библиотеки
import pandas as pd
import numpy as np

print('Pandas версия:', pd.__version__)
print('Готово!')
````

## Часть 1: Series — одна кривая каротажа

````python
# Series — это как столбец в Excel с индексом
# Создаём кривую гамма-каротажа (ГК)
GK = pd.Series(
    data  = [45, 62, 78, 55, 90, 110, 88, 42, 67, 73],
    index = [1000, 1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009],
    name  = 'GR_API'
)

print(GK)
print()
print('Тип:', type(GK))
print('Среднее ГК:', GK.mean().round(1))
````

## Часть 2: DataFrame — полный планшет скважины

DataFrame = несколько Series объединённых в одну таблицу.

````python
# Создаём таблицу данных скважины
data = {
    'Глубина_м':  [1000, 1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009],
    'ГК_API':     [45,   62,   78,   55,   90,   110,  88,   42,   67,   73],
    'RHOB_г_см3': [2.65, 2.70, 2.45, 2.42, 2.38, 2.50, 2.68, 2.71, 2.60, 2.55],
    'NPHI_доли':  [0.12, 0.14, 0.28, 0.30, 0.33, 0.25, 0.13, 0.11, 0.18, 0.22],
    'Литология':  ['Известняк','Известняк','Песчаник','Песчаник',
                   'Песчаник','Песчаник','Известняк','Известняк',
                   'Глина','Глина']
}

df = pd.DataFrame(data)
print(df)
````

## Часть 4: Выбор столбцов и строк

### 4.1 Выбор столбцов

````python
# Один столбец
print(df['ГК_API'])
````

````python
# Несколько столбцов
df[['Глубина_м', 'ГК_API', 'Литология']]
````

### 4.2 Выбор строк

| Метод | Что делает |
|-------|------------|
| `.iloc[i]` | По номеру строки (как индекс списка) |
| `.loc[i]` | По метке индекса |

````python
# iloc — по номеру (0 = первая строка)
print('Первая строка:')
print(df.iloc[0])
print()
print('Строки 2-4:')
print(df.iloc[2:5])
````

---
## Часть 5: Фильтрация — находим нужные пласты

Это одна из самых важных операций в анализе данных!

````python
# Найдём все строки где ГК < 65 (потенциальные коллекторы)
kollektory = df[df['ГК_API'] < 65]
print('Потенциальные коллекторы (ГК < 65 API):')
print(kollektory)
````

````python
# Только песчаники
peschaniki = df[df['Литология'] == 'Песчаник']
print('Все песчаники:')
print(peschaniki)
````

````python
# Несколько условий — & (И), | (ИЛИ)
filter_mask = (df['Литология'] == 'Песчаник') & (df['RHOB_г_см3'] < 2.50)
poristy = df[filter_mask]
print('Пористые песчаники (RHOB < 2.50):')
print(poristy)
````

## Шпаргалка дня

```python
import pandas as pd

# Создание
df = pd.DataFrame({'col1': [1,2], 'col2': [3,4]})
df = pd.read_csv('file.csv')

# Разведка
df.shape          # размер
df.head()         # первые 5 строк
df.info()         # типы и пропуски
df.describe()     # статистика

# Выбор
df['col']         # один столбец
df[['c1','c2']]   # несколько столбцов
df.iloc[0]        # строка по номеру

# Фильтрация
df[df['col'] > 10]
df[(df['col1'] > 10) & (df['col2'] < 5)]
df[df['col'] == 'текст']
```
